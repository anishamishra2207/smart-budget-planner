```python
from IPython.display import Markdown, display
from agent import BudgetAgent

display(Markdown("## Smart Budget Planner Agent"))
display(Markdown("Initialize one in-memory agent. Its state and conversation history will persist across the next three turns."))
agent = BudgetAgent(total_budget=1000)
print("Agent initialized with total budget: $1000.00")
```

```python
display(Markdown("### Turn 1: Record expenses"))
user_message = 'I spent 200 on textbooks and 50 on snacks.'
print(f"User: {user_message}")
answer = agent.run_agent_turn(user_message)
print(f"Assistant: {answer}")
```

```python
display(Markdown("### Turn 2: Query current budget"))
user_message = 'How much budget do I have left?'
print(f"User: {user_message}")
answer = agent.run_agent_turn(user_message)
print(f"Assistant: {answer}")
```

```python
display(Markdown("### Turn 3: Test affordability reasoning"))
user_message = 'Can I afford a 900 trip?'
print(f"User: {user_message}")
answer = agent.run_agent_turn(user_message)
print(f"Assistant: {answer}")
display(Markdown("The printed `[Step n Action]` and `[Step n Observation]` lines show the agent querying or updating memory before producing each answer."))
```
